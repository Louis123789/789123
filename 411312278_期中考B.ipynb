````python
# ============================================================
# 第 2 格：建立資料庫與初始資料
# 提醒：每次執行都要 DROP TABLE IF EXISTS 再重建
# ============================================================
import sqlite3

def init_db():
    conn = sqlite3.connect('coffee_shop.db')
    cursor = conn.cursor()

    # 每次執行都先刪除舊資料表，確保乾淨重建
    cursor.execute('DROP TABLE IF EXISTS orders')
    cursor.execute('DROP TABLE IF EXISTS menu')

    # 建立菜單資料表
    cursor.execute('''
        CREATE TABLE menu (
            item_id INTEGER PRIMARY KEY AUTOINCREMENT,
            item_name TEXT UNIQUE NOT NULL,
            price INTEGER NOT NULL,
            stock INTEGER NOT NULL
        )
    ''')

    # 建立訂單資料表
    cursor.execute('''
        CREATE TABLE orders (
            order_id INTEGER PRIMARY KEY AUTOINCREMENT,
            student_name TEXT NOT NULL,
            item_name TEXT NOT NULL,
            quantity INTEGER NOT NULL,
            total_price INTEGER NOT NULL,
            status TEXT NOT NULL
        )
    ''')

    # 放入初始資料
    initial_menu = [
        ('美式咖啡', 50, 5),
        ('拿鐵', 65, 4),
        ('抹茶歐蕾', 70, 3)
    ]
    cursor.executemany('''
        INSERT INTO menu (item_name, price, stock)
        VALUES (?, ?, ?)
    ''', initial_menu)

    conn.commit()
    conn.close()

init_db()
print("✅ 資料庫與初始資料建立完成！")
````

````python
# ============================================================
# 第 3 格：顯示資料表
# 提醒：定義 show_tables() 並呼叫一次
# ============================================================
import pandas as pd

def show_tables():
    conn = sqlite3.connect('coffee_shop.db')
    try:
        print("=== 菜單資料表 (menu) ===")
        df_menu = pd.read_sql_query("SELECT * FROM menu", conn)
        display(df_menu)

        print("\n=== 訂單資料表 (orders) ===")
        df_orders = pd.read_sql_query("SELECT * FROM orders", conn)
        display(df_orders)
    finally:
        conn.close()

show_tables()
````

````python
# ============================================================
# 第 6 格：功能函式
# 提醒：fetchone() 後先檢查 None；用 try/finally 關閉連線；統計用 JOIN＋GROUP BY
# ============================================================
def process_order(student_name, item_name, quantity):
    if not student_name or not student_name.strip():
        return "❌ 訂購失敗：請輸入學生姓名。"

    if quantity < 1 or quantity > 3:
        return "❌ 訂購失敗：每次訂購數量需在 1 至 3 杯之間！"

    conn = sqlite3.connect('coffee_shop.db')
    cursor = conn.cursor()
    try:
        # 檢查個人今日累計訂購量（已取消的不計）
        cursor.execute('''
            SELECT SUM(quantity) FROM orders
            WHERE student_name = ? AND status = '已完成'
        ''', (student_name,))
        row = cursor.fetchone()
        current_total = row[0] if row and row[0] is not None else 0

        if current_total + quantity > 3:
            return f"❌ 訂購失敗：每人每日限購 3 杯（您今日已訂購 {current_total} 杯，欲加購 {quantity} 杯將超過上限）。"

        # 檢查品項庫存與單價
        cursor.execute('SELECT price, stock FROM menu WHERE item_name = ?', (item_name,))
        menu_row = cursor.fetchone()
        if not menu_row:
            return f"❌ 訂購失敗：找不到品項「{item_name}」。"

        price, stock = menu_row[0], menu_row[1]
        if stock < quantity:
            return f"❌ 訂購失敗：{item_name} 庫存不足（剩餘 {stock} 杯）。"

        total_price = price * quantity

        # 扣減庫存
        cursor.execute('UPDATE menu SET stock = stock - ? WHERE item_name = ?', (quantity, item_name))

        # 新增訂單紀錄
        cursor.execute('''
            INSERT INTO orders (student_name, item_name, quantity, total_price, status)
            VALUES (?, ?, ?, ?, '已完成')
        ''', (student_name, item_name, quantity, total_price))

        conn.commit()
        return f"✅ 訂購成功！{student_name} 訂購 {item_name} {quantity} 杯，共 {total_price} 元。"
    finally:
        conn.close()

def process_cancel(student_name, item_name):
    if not student_name or not student_name.strip():
        return "❌ 取消失敗：請輸入學生姓名。"

    conn = sqlite3.connect('coffee_shop.db')
    cursor = conn.cursor()
    try:
        # 尋找該學生該品項最新一筆「已完成」的訂單
        cursor.execute('''
            SELECT order_id, quantity FROM orders
            WHERE student_name = ? AND item_name = ? AND status = '已完成'
            ORDER BY order_id DESC LIMIT 1
        ''', (student_name, item_name))
        order_row = cursor.fetchone()

        if not order_row:
            return f"❌ 取消失敗：找不到您預約「{item_name}」的有效訂單。"

        order_id, quantity = order_row[0], order_row[1]

        # 更新訂單狀態為已取消
        cursor.execute("UPDATE orders SET status = '已取消' WHERE order_id = ?", (order_id,))

        # 加回庫存
        cursor.execute("UPDATE menu SET stock = stock + ? WHERE item_name = ?", (quantity, item_name))

        conn.commit()
        return f"✅ 已成功取消 {item_name} {quantity} 杯，庫存已釋出。"
    finally:
        conn.close()

def process_query(item_name):
    conn = sqlite3.connect('coffee_shop.db')
    cursor = conn.cursor()
    try:
        cursor.execute("SELECT stock FROM menu WHERE item_name = ?", (item_name,))
        row = cursor.fetchone()
        if not row:
            return f"❌ 查無此品項：「{item_name}」。"
        return f"ℹ️ {item_name} 目前剩餘庫存為 {row[0]} 杯。"
    finally:
        conn.close()

def process_statistics():
    conn = sqlite3.connect('coffee_shop.db')
    try:
        query = """
            SELECT
                m.item_name AS 品項,
                m.price AS 單價,
                m.stock AS 剩餘庫存,
                COALESCE(SUM(CASE WHEN o.status = '已完成' THEN o.quantity ELSE 0 END), 0) AS 已售杯數,
                COALESCE(SUM(CASE WHEN o.status = '已完成' THEN o.total_price ELSE 0 END), 0) AS 營收
            FROM menu m
            LEFT JOIN orders o ON m.item_name = o.item_name
            GROUP BY m.item_name, m.price, m.stock
        """
        df = pd.read_sql_query(query, conn)
        return df
    finally:
        conn.close()

def handle_user_request(student_name, user_text):
    action, item, quantity, raw = parse_user_input(user_text)

    if action == "訂購":
        return process_order(student_name, item, quantity)
    elif action == "取消":
        return process_cancel(student_name, item)
    elif action == "查詢":
        return process_query(item)
    elif action == "統計":
        df_stats = process_statistics()
        return df_stats
    else:
        return f"🤖 抱歉，無法識別您的需求。目前提供品項：美式咖啡、拿鐵、抹茶歐蕾。"

print("✅ 功能函式載入完成")
````
